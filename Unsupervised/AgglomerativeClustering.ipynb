```python
import numpy as np
import matplotlib.pyplot as plt
import mglearn

from sklearn.cluster import AgglomerativeClustering
```

```python
from sklearn.datasets import make_blobs

X, y = make_blobs(
    n_samples=100,
    random_state=42
)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=y
)

plt.xlabel("Feature 0")
plt.ylabel("Feature 1")
plt.title("Dataset")
plt.show()
```

```python
agg = AgglomerativeClustering(
    n_clusters=3
)

print(agg)
```

```python
assignment = agg.fit_predict(X)

print("Hasil clustering:")
print(assignment)
```

```python
print("Jumlah cluster:", len(np.unique(assignment)))

print("Cluster yang terbentuk:")
print(np.unique(assignment))
```

```python
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=assignment
)

plt.xlabel("Feature 0")
plt.ylabel("Feature 1")
plt.title("Agglomerative Clustering")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(
    X[:, 0],
    X[:, 1],
    c=y
)
axes[0].set_title("Data Asli")
axes[0].set_xlabel("Feature 0")
axes[0].set_ylabel("Feature 1")

axes[1].scatter(
    X[:, 0],
    X[:, 1],
    c=assignment
)
axes[1].set_title("Agglomerative Clustering")
axes[1].set_xlabel("Feature 0")
axes[1].set_ylabel("Feature 1")

plt.show()
```
